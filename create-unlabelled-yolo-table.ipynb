# Active Labeling with YOLO and 3LC

Prerequisites:
+ A trained model
+ Existing dataset(s) / 3LC Table(s)
+ Folder(s) of unlabelled images

Objectives:
+ Label all our images
+ Label a subset of images (by relevance)

```python
import tlc

from tlc_ultralytics import YOLO, Settings
```

```python
PROJECT_NAME = "YOLO ACTIVE LEARNING"
```

## Create a YOLO  yaml file for the unlabeled data

Should contain `path`, `train` and `names`. `names` should be the same as the reference table.

## Create a TableFromYOLO using the unlabeled data yaml file

```python
unlabeled_table = tlc.Table.from_yolo(
    dataset_yaml_file="C:/Project/3lc-ultralytics/unlabelled_yolo.yaml",
    split="train",
    table_name="unlabeled_yolo_table",
    weight_column_value=1,  # 0 is the weight for the unlabeled data
    project_name=PROJECT_NAME,
)
```

```python
unlabeled_table[0]
```

## Load a labeled table

```python
labeled_table = tlc.Table.from_yolo(
    dataset_yaml_file="C:/Project/ultralytics/ultralytics/cfg/datasets/coco128.yaml",
    split="train",
    table_name="labeled_table",
    project_name="YOLO ACTIVE LEARNING",
)
```

```python
labeled_table[0]
```

```python

model = YOLO("yolo11n.pt")  # Or load your current best model weights

results = model.collect(
    
)
```

## Do active labeling in the 3LC Dashboard

E.g. accept confident predictions and set weight to 1.0 for a subset of the unlabeled data.

## Get the latest revision of the unlabeled table

```python
latest_unlabeled_table = unlabeled_table.latest()
```

```python
latest_unlabeled_table
```

## Join the labeled and unlabeled tables to create a new training table

```python
joined_train_table = tlc.Table.join_tables([labeled_table, latest_unlabeled_table])
```

## Train a new model (exclude 0-weight samples, collect on all samples)

```python
model.train(
    tables={
        "train": joined_train_table.latest(),
        "val": joined_train_table,
    },
    settings=Settings(
        image_embeddings_dim=2,
        collect_loss=True,
    ),
    workers=0,
    verbose=False,
    epochs=10,
)
```

```python
tlc.active_run()
```
